# Data types
Below are some common data types in python

## Numeric data types:
### int : Represents the whole numbers e.g. 10, -5, 0
### float : Represents the decimal numbers eg. 3.14, -0.5, 2.0
### complex : Represents the complex numbers with real and imaginary part e.g. 1+2j

## Text Type:
### str: Represents the sequence of characters enclosed in single or double quotes. e.g 'Hello', "Python"

## Boolean types:
### bool : Represents the truth values, either true or false

## Sequence Types:
### list : An ordered, mutable collection of items, enclosed in square brackets e.g. [1,2,3], ['Apple', 'Banana', 'Chickoo']
### tuple : An ordered, immutable collection of items, enclosed in round brackets/parantheses. eg. (1,2,3), ("red", "yellow", "green")
### range : Range represents the immutable sequence of numbers often used in loops. eg. range(5)

## Mapping Types:
### dict : An unordered collection of key-value pairs, enclosed in curly braces. {'name': 'Alice', 'age': 30}

## Set Types:
### set: An unordered collection of unique, mutable items, enclosed in curly braces. {1, 2, 3}, {'a', 'b', 'c'}

## Binary Types: 
### bytes: An immutable sequence of bytes.
### bytearray: A mutable sequence of bytes.
### memoryview: Provides a memory-efficient way to access the internal data of an object.

## None Types:
### NoneType : Represents  the absence of a value, with the single value None.

```python
# Switch case

day = "Monday"

match day.upper():
    case "MONDAY" | "TUESDAY" | "WEDNESDAY" | "THURSDAY" | "FRIDAY":
        print("It's a weekday.")
    case "SATURDAY" | "SUNDAY":
        print("It's the weekend.")
    case _:
        print("Not a valid day.")
```

```python
# loops

# While loop
i = 1
while i < 6:
    print(i)
    if i == 3:
        break
    i += 1

# For loop
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    if fruit == "banana":
        break
    print(fruit)


for x in "banana":
    print(x)
```

```python
# Functions / Methods: Funciton are defined using def keyword in Python. Python is indentation based laguage. So indentation is very important.
def greet(): 
    print("Hello World!")

greet()

# passing parameters to functions
def greet_user(username):
    print("Hello, " + username + "!")

greet_user("Alice")

# passing multiple parameters
def add_numbers(a, b):
    return a + b

result = add_numbers(5, 10)
print("Sum is:", result)


# arbirtrary number of arguments
def sum_all(*args):
    total = 0
    for num in args:
        total += num
    return total

total_sum = sum_all(1, 2, 3, 4, 5)
print("Total Sum is:", total_sum)


# arbitary keyword arguments
def print_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")


print_info(name="Alice", age=30, city="New York")

```
